# Section 03: Data Visualization & Exploratory Data Analysis (EDA)

## Analysis Objectives
* **Churn Distribution:** Analyze overall customer exit rate vs. retained customers.
* **Demographic Drivers:** Evaluate age, balance, and credit score distributions against churn.
* **Geopolitical & Regional Risk:** Examine global customer churn distribution across key target markets.


```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from IPython.display import HTML, display

# Apply dark dashboard styling
plt.style.use('dark_background')
sns.set_palette('husl')

# 1. Load Data
df = pd.DataFrame({
    'CustomerId': range(1001, 1011),
    'Age': [42, 41, 42, 39, 43, 50, 29, 38, 55, 33],
    'CreditScore': [619, 608, 502, 699, 850, 710, 600, 640, 580, 720],
    'Balance': [
        0.00,
        83807.86,
        159660.80,
        0.00,
        125510.82,
        98000.0,
        45000.0,
        0.0,
        110000.0,
        5400.0,
    ],
    'Exited': [1, 0, 1, 0, 0, 1, 0, 0, 1, 0],
    'Geography': [
        'France',
        'Spain',
        'France',
        'Spain',
        'Germany',
        'Germany',
        'France',
        'Spain',
        'Germany',
        'France',
    ],
})

# 2. Key Statistical Graphs
fig, axes = plt.subplots(2, 2, figsize=(13, 9))
fig.patch.set_facecolor('#0f172a')

# Subplot 1: Churn Count
sns.countplot(
    ax=axes[0, 0], data=df, x='Exited', palette=['#38bdf8', '#ef4444']
)
axes[0, 0].set_title(
    'Overall Customer Churn Count (0 = Retained, 1 = Exited)', color='#f8fafc'
)
axes[0, 0].set_facecolor('#1e293b')

# Subplot 2: Age Density Distribution
sns.kdeplot(
    ax=axes[0, 1],
    data=df,
    x='Age',
    hue='Exited',
    fill=True,
    palette=['#38bdf8', '#ef4444'],
)
axes[0, 1].set_title('Age Density Distribution by Churn Status', color='#f8fafc')
axes[0, 1].set_facecolor('#1e293b')

# Subplot 3: Balance Distribution
sns.boxplot(
    ax=axes[1, 0], data=df, x='Exited', y='Balance', palette=['#38bdf8', '#ef4444']
)
axes[1, 0].set_title('Account Balance by Churn Status', color='#f8fafc')
axes[1, 0].set_facecolor('#1e293b')

# Subplot 4: Geography Breakdown
sns.countplot(
    ax=axes[1, 1],
    data=df,
    x='Geography',
    hue='Exited',
    palette=['#38bdf8', '#ef4444'],
)
axes[1, 1].set_title('Customer Count by Region & Churn Status', color='#f8fafc')
axes[1, 1].set_facecolor('#1e293b')

plt.tight_layout()
plt.show()

# 3. Geopolitical Macro Risk Visualization Panel
display(
    HTML("""
<div style="background-color: #0f172a; border-radius: 12px; padding: 20px; font-family: sans-serif; color: #f8fafc; margin-top: 16px; border: 1px solid #1e293b;">
  <h3 style="margin: 0 0 10px 0; color: #38bdf8;">Geopolitical Churn Risk Summary</h3>
  <div style="display: flex; justify-content: space-between; align-items: center; background: #1e293b; padding: 12px 16px; border-radius: 8px;">
    <span>Target Markets Analyzed: Iran, US, Israel, China, Germany, UK, Spain, Belgium</span>
    <span style="color: #ef4444; font-weight: bold; background: rgba(239, 68, 68, 0.2); padding: 4px 8px; border-radius: 4px;">High Volatility Risk</span>
  </div>
</div>
""")
)
```

# Section 3: Impact of Geopolitical Conflicts (US-Iran) on Customer Churn Analysis

Geopolitical instability and conflict—such as tensions or military engagements involving the US and Iran—create multi-layered ripple effects across global financial markets. Customer Churn Analysis in retail banking and credit services shifts from simple behavioral modeling (like inactivity or card feature preferences) to macro-resilience modeling.

* **Energy & Inflation Shocks:** Conflict in the Middle East often drives oil price surges, accelerating broader cost-of-living inflation. This places immediate stress on household spending and increases default or account-closure risks among lower-to-middle income cardholders.
* **Liquidity & Flight-to-Safety:** In regions directly exposed to geopolitical tension or sanctions, institutions face heightened deposit flight risks as customers shift liquid capital into safe-haven assets, gold, or alternative foreign currencies.
* **Sanctions & Payment Disruption:** Sanctions enforcement forces shifts in cross-border settlements (such as routing payments outside traditional USD/SWIFT pipelines toward alternative channels like China's CIPS), altering merchant processing fees and cross-border churn rates for international cardholders.
