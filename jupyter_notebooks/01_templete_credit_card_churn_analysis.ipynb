# Section 1 - Exploratory Data Analysis (EDA)

* Load customer demographic and transaction dataset.
* Analyze churn patterns using distributions and behavioral feature comparisons.

Section 1 content

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style matching clean analytics standards
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.sans-serif': 'DejaVu sans', 'font.size': 10})

# Synthetic data Generation (Replace with pd.read_csv if loading external CSV)
np.random.seed(42) # For reproductibility
n_samples = 1000
n_samples = 1000

# Create data
df = pd.DataFrame({
    'Age': np.random.randint(20, 70, n_samples), 
    'Gender': np.random.choice(['Female', 'Male'], n_samples, p=[0.52, 0.48]),
    'CreditScore': np.random.randint(500, 850, n_samples),
    'Total_Trans_Amt': np.random.exponential(scale=3000, size=n_samples) + 1000,
    'Total_Trans_Ct': np.random.randint(20, 130, n_samples),
    'Churn': np.random.choice([0, 1], n_samples, p=[0.80, 0.20])
})

#Create 4-panel Figure
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Credit Card Customer Churn Analysis: Demographics & Transactions', fontsize=16, fontweight='bold', y=0.98)

# Chart 1: Age Distribution
sns.histplot(data=df, x='Age',hue='Churn', kde=True, palette=['#3B82F6', '#EF4444'], ax=axes[0, 0], element="step")
axes[0, 0,].set_title('Age Distribution by Churn Status', fontweight='bold')

#Chart 2: Churn Rate by Gender
sns.barplot(data=df, x='Gender', y='Churn', palette=['#93C5FD', '#FCA5A5'], ax=axes[0, 1], errorbar=None)
axes[0, 1]. set_title('Churn Rate by Gender', fontweight='bold')
axes[0, 1].yaxis.set_major_formatter('{x:.0%}')

# Chart 3: Amount vs Count Scatter
sns.scatterplot(data=df, x='Total_Trans_Ct', y='Total_Trans_Amt', hue='Churn', palette=['#3B82F6', '#EF4444'], alpha=0.7, ax=axes[1, 0])
axes[1, 0].set_title('Transaction Amount vs. Count', fontweight='bold')

# Chart 4: Credit Score Boxplot
sns.boxplot(data=df, x='Churn', y='CreditScore', palette=['#93C5FD', '#FCA5A5'], ax=axes[1, 1])
axes[1, 1].set_title('Credit Score Distribution by Churn Status', fontweight='bold')
axes[1, 1].set_xticklabels(['Retained (0)', 'Churned (1)'])

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

Impact of Geopolitical Conflicts (US-Iran) on Customer Churn Analysis
Geopolitical instability and conflict—such as tensions or military engagements involving the US and Iran—create multi-layered ripple effects across global financial markets. Customer Churn Analysis in retail banking and credit services shifts from simple behavioral modeling (like inactivity or card feature preferences) to macro-resilience modeling.

Key Drivers of Geopolitical Customer Churn
⚬	Energy & Inflation Shocks: Conflict in the Middle East often drives oil price surges, accelerating broader cost-of-living inflation. This places immediate stress on household spending and increases default or account-closure risks among lower-to-middle income cardholders.
⚬	Liquidity & Flight-to-Safety: In regions directly exposed to geopolitical tension or sanctions, institutions face heightened deposit flight risks as customers shift liquid capital into safe-haven assets, gold, or alternative foreign currencies.
⚬	Sanctions & Payment Disruption: Sanctions enforcement forces shifts in cross-border settlements (such as routing payments outside traditional USD/SWIFT pipelines toward alternative channels like China's CIPS), altering merchant processing fees and cross-border churn rates for international cardholders.

By systematically addressing these core risk factors, financial institutions can shift from reactive retention to proactive, high-value customer engagement.
Key Strategic Insights

⚬	Macro & Geopolitical Volatility: Regional market disruptions, currency shifts, and economic pressures (such as inflation shocks and cross-border fee changes) accelerate non-traditional churn across global cardholder segments.
Recommended Action Plan
⚬	Automate Early-Warning Systems: Deploy predictive machine learning models to trigger automated retention offers (e.g., fee waivers, cashback boosts) when customer activity metrics drop below risk thresholds.
